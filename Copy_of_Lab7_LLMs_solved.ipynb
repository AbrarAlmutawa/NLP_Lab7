# Part A — Decoder-Only Model

## Decoder-only example: GPT-style Text Generation

A decoder-only model predicts the next token based on previous tokens.

We will use **DistilGPT-2**, a small GPT-style model.


````python
generator = pipeline("text-generation", model="distilgpt2")

prompt = "Natural language processing is"

output = generator(prompt, max_new_tokens=30, do_sample=False)

print(output[0]["generated_text"])
````

# Part B — Encoder–Decoder Model

## Encoder–decoder example: Instruction / Text-to-Text Generation

Encoder–decoder models contain two main parts:

1. The **encoder** reads the input.
2. The **decoder** produces the output.

We will use **FLAN-T5-small**.


````python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

tokenizer = AutoTokenizer.from_pretrained("google/flan-t5-small")
model = AutoModelForSeq2SeqLM.from_pretrained("google/flan-t5-small")

prompt = "Translate to French: I love natural language processing."

# Convert text into tokens
output = model.generate(**tokenizer(prompt, return_tensors="pt"))

# Convert tokens back to text
print(tokenizer.decode(output[0], skip_special_tokens=True))
````

# Part C — Encoder-Only Model

## Encoder-only example: Sentiment Classification

Encoder-only models read the full input and create a representation of it.

They are very useful when the task is to **understand or classify text**.

Here we use a DistilBERT sentiment model.


````python
classifier = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")

text = "This NLP lab is easy and interesting."

result = classifier(text)
print(result)
````

## Task 3 — Encoder–Decoder Model

Use the FLAN-T5 model to perform the following:

- Summarize a short paragraph.


````python
paragraph = (
    "Natural language processing is a field of artificial intelligence that helps computers "
    "understand, interpret, and generate human language. It is used in many everyday applications "
    "such as machine translation, chatbots, spam filtering, and voice assistants. Modern NLP systems "
    "rely on transformer models, which learn patterns in language from very large amounts of text."
)

prompt = "Summarize the following text in one sentence: " + paragraph

inputs = tokenizer(prompt, return_tensors="pt")
summary_ids = model.generate(**inputs, max_new_tokens=40, num_beams=4)

print("Original paragraph:\n", paragraph)
print("\nSummary:\n", tokenizer.decode(summary_ids[0], skip_special_tokens=True))
````


## Task 5 — Short Reflection

Answer in 3–5 sentences:

**Why is a decoder-only model suitable for a chatbot such as ChatGPT?**


**Answer:**

A decoder-only model is trained to predict the next token given all the previous tokens, which is exactly what a chatbot does: it reads the conversation so far and continues it with a reply. Because it generates text one token at a time from left to right, it can produce open-ended responses of any length rather than just a label or a fixed output. The whole conversation (user messages and earlier replies) can be fed in as one long prompt, so the model uses that context to keep its answers coherent. Finally, decoder-only models scale very well when trained on huge amounts of text, which gives them broad knowledge and fluent language, the main qualities a general-purpose assistant like ChatGPT needs.

### Question

Why do you sometimes get different outputs even when the prompt is the same?

**Hint:** Sampling introduces randomness when choosing the next token.


````python
# Sampling WITHOUT set_seed(42): run this cell several times
prompt = "Natural language processing is"

for i in range(3):
    out = generator(prompt, max_new_tokens=30, do_sample=True, top_k=50, temperature=0.9)
    print(f"Run {i+1}: {out[0]['generated_text']}\n")
````

**Answer:** With `do_sample=True`, the model does not always pick the single most likely next token. Instead, it computes a probability distribution over the vocabulary and randomly *samples* a token from it (here limited to the top 50 candidates and shaped by the temperature). Since a different token can be chosen at any step, and every later token depends on the earlier ones, the whole continuation can change from run to run even with the same prompt. `set_seed(42)` fixes the random number generator so the same "random" choices are made every time; removing it makes each run different. In contrast, `do_sample=False` (greedy decoding, used in Part A) always picks the most likely token, so it gives the same output every time.
